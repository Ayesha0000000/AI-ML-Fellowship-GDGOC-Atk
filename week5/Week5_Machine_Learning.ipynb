## Introduction to Machine Learning

Machine Learning (ML) is a branch of Artificial Intelligence that allows computers to learn from data without being explicitly programmed.

Instead of writing rules manually, we give data to the machine, and it finds patterns and makes predictions.

### Types of Machine Learning

1. Supervised Learning  
   The model learns from labeled data.  
   Example: Predicting house prices.

2. Unsupervised Learning  
   The model finds patterns in unlabeled data.  
   Example: Customer grouping.

3. Reinforcement Learning  
   The model learns through rewards and punishments.  
   Example: Game playing AI.

### Real Life Examples

- Netflix recommendations  
- Email spam detection  
- Face recognition  
- Voice assistants like Siri

```python
# Import Libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Sample data
X = np.array([1,2,3,4,5])
Y = np.array([2,4,5,4,5])
```

## Linear Regression From Scratch

Linear Regression is used to find a relationship between two variables.

Equation:

y = mx + b

Where:
m = slope  
b = intercept

```python
mean_x = np.mean(X)
mean_y = np.mean(Y)
```

```python
numerator = np.sum((X - mean_x)*(Y - mean_y))
denominator = np.sum((X - mean_x)**2)

m = numerator / denominator
b = mean_y - m * mean_x

print("Slope:", m)
print("Intercept:", b)
```

```python
Y_pred = m * X + b
```

```python
plt.scatter(X, Y, color="blue")
plt.plot(X, Y_pred, color="red")
plt.title("Linear Regression From Scratch")
plt.show()
```

## Linear Regression Using Scikit-Learn

Now we will use built-in library to perform Linear Regression.

```python
from sklearn.linear_model import LinearRegression

X_new = X.reshape(-1,1)

model = LinearRegression()
model.fit(X_new, Y)

Y_pred2 = model.predict(X_new)

plt.scatter(X, Y)
plt.plot(X, Y_pred2, color="green")
plt.title("Linear Regression using Sklearn")
plt.show()
```

## Gradient Descent Implementation

Gradient Descent is an optimization algorithm used to minimize the error in prediction.

It updates slope (m) and intercept (b) step by step until the error becomes very small.

```python
m = 0
b = 0
learning_rate = 0.01
iterations = 1000

n = len(X)

for i in range(iterations):
    Y_pred = m*X + b

    dm = (-2/n) * np.sum(X*(Y - Y_pred))
    db = (-2/n) * np.sum(Y - Y_pred)

    m = m - learning_rate * dm
    b = b - learning_rate * db

print("Final slope:", m)
print("Final intercept:", b)
```

```python
Y_pred = m*X + b

plt.scatter(X, Y)
plt.plot(X, Y_pred, color="purple")
plt.title("Gradient Descent Result")
plt.show()
```
