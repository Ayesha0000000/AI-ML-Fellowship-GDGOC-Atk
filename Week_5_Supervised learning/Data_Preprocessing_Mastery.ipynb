# Task 1: Handling Missing Data – Titanic Dataset

## Task Overview
The objective of this task is to identify and handle missing values in the Titanic dataset.

## Techniques Used
- Median Imputation (Age)
- Mode Imputation (Embarked)
- Dropping Column (Cabin)

## Why These Techniques?
- Age is numerical → Median is robust to outliers.
- Embarked is categorical → Mode is appropriate.
- Cabin contains excessive missing values → Dropping avoids noise.

```python
import pandas as pd

df = pd.read_csv("train (1).csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df.drop('Cabin', axis=1, inplace=True)
```

```python
df.isnull().sum()
```

## Results & Observations

- Missing values were identified in Age, Cabin, and Embarked.
- Age was filled using median imputation.
- Embarked was filled using mode imputation.

- The dataset now contains no missing values and is ready for modeling.

# Task 4: Handling Outliers – Boston Housing Dataset

## Task Overview
The objective of this task is to detect and treat outliers in the Boston Housing dataset.

## Techniques Used
- Z-Score Method
- IQR Method

## Why These Techniques?
- Z-Score identifies extreme values based on standard deviation.
- IQR method detects outliers using interquartile range.
- Removing outliers improves model stability and reduces noise.

```python
from sklearn.datasets import fetch_openml
import pandas as pd

boston = fetch_openml(name="boston", version=1, as_frame=True)
df_boston = boston.frame

df_boston.head()
```

```python
import numpy as np
from scipy import stats

# Sirf numerical columns select karein
df_numeric = df_boston.select_dtypes(include=[np.number])

# Z-score calculate karein
z_scores = np.abs(stats.zscore(df_numeric))

# Outliers remove karein
df_zscore = df_boston[(z_scores < 3).all(axis=1)]

print("Original Shape:", df_boston.shape)
print("After Z-Score Removal:", df_zscore.shape)
```

```python
Q1 = df_numeric.quantile(0.25)
Q3 = df_numeric.quantile(0.75)
IQR = Q3 - Q1

df_iqr = df_boston[~((df_numeric < (Q1 - 1.5 * IQR)) |
                      (df_numeric > (Q3 + 1.5 * IQR))).any(axis=1)]

print("Original Shape:", df_boston.shape)
print("After IQR Removal:", df_iqr.shape)
```

## Results & Observations

- Z-Score method removed data points beyond 3 standard deviations.
- IQR method removed values outside the interquartile range.
- The dataset size decreased after outlier removal.
- Removing outliers helps reduce noise and improves model stability.

# Task 5: Advanced Data Imputation – Retail Sales Dataset

## Task Overview
The objective of this task is to handle missing values using advanced imputation techniques.

## Techniques Used
- KNN Imputation
- MICE (Iterative Imputer)

## Why These Techniques?
- KNN fills missing values using nearest neighbors.
- MICE uses regression-based iterative estimation.
- Advanced imputation provides more realistic value estimation than simple mean/median.

```python
!wget https://raw.githubusercontent.com/selva86/datasets/master/RetailMarketing.csv
```

```python
import pandas as pd
import numpy as np

# Simple retail-like dataset
np.random.seed(42)

df_retail = pd.DataFrame({
    "Sales": np.random.randint(100, 1000, 200),
    "Customers": np.random.randint(10, 100, 200),
    "Discount": np.random.uniform(0, 0.5, 200),
    "Inventory": np.random.randint(50, 500, 200)
})

df_retail.head()
```

```python
df_retail_missing = df_retail.copy()

for col in df_retail_missing.columns:
    df_retail_missing.loc[df_retail_missing.sample(frac=0.1).index, col] = np.nan

df_retail_missing.isnull().sum()
```

```python
from sklearn.impute import KNNImputer

knn_imputer = KNNImputer(n_neighbors=5)

df_knn = pd.DataFrame(
    knn_imputer.fit_transform(df_retail_missing),
    columns=df_retail_missing.columns
)

df_knn.isnull().sum()
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

mice_imputer = IterativeImputer(random_state=42)

df_mice = pd.DataFrame(
    mice_imputer.fit_transform(df_retail_missing),
    columns=df_retail_missing.columns
)

df_mice.isnull().sum()
```

## Results & Observations

- Artificial missing values were introduced.
- KNN Imputer filled values using nearest neighbors.
- MICE used regression-based iterative estimation.
- Both methods successfully removed missing values.
- Advanced imputation provides more realistic estimates than simple mean or median.

## Techniques Used
- Log Transformation
- Square Root Transformation
- Box-Cox Transformation

## Why?
Transformations reduce skewness and stabilize variance,
which improves performance of machine learning models.

```python
from sklearn.datasets import load_diabetes
import pandas as pd

data = load_diabetes(as_frame=True)
df_bike = data.frame   # naam same rakh rahe hain consistency ke liye

df_bike.head()
```

```python
df_bike["target_sqrt"] = np.sqrt(df_bike["target"] + shift_value)

df_bike[["target", "target_sqrt"]].head()
```

```python
print("Original Skewness:", df_bike["target"].skew())
print("Log Skewness:", df_bike["target_log"].skew())
print("Sqrt Skewness:", df_bike["target_sqrt"].skew())
print("BoxCox Skewness:", df_bike["target_boxcox"].skew())
```

## Results & Observations

- The original variable showed skewness.

# Task 9: Handling Imbalanced Data – Credit Card Fraud

## Task Overview
The objective of this task is to handle class imbalance in a dataset.

## Techniques Used
- Random Undersampling
- SMOTE
- ADASYN

## Why?
Imbalanced datasets bias models toward majority class.
Balancing improves minority class prediction.

```python
from sklearn.datasets import make_classification
import pandas as pd

X, y = make_classification(
    n_samples=5000,
    n_features=10,
    n_classes=2,
    weights=[0.95, 0.05],   # 95% majority, 5% minority
    random_state=42
)

df_imb = pd.DataFrame(X)
df_imb["target"] = y

df_imb["target"].value_counts()
```

```python
X = df_imb.drop("target", axis=1)
y = df_imb["target"]
```

```python
from imblearn.under_sampling import RandomUnderSampler

rus = RandomUnderSampler(random_state=42)
X_under, y_under = rus.fit_resample(X, y)

pd.Series(y_under).value_counts()
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)
X_smote, y_smote = smote.fit_resample(X, y)

pd.Series(y_smote).value_counts()
```

```python
from imblearn.over_sampling import ADASYN

adasyn = ADASYN(random_state=42)
X_ada, y_ada = adasyn.fit_resample(X, y)

pd.Series(y_ada).value_counts()
```

## Results & Observations

- Original dataset was highly imbalanced.
- Random undersampling reduced majority class size.
- SMOTE generated synthetic minority samples.
- ADASYN adaptively generated minority samples.
- Balanced datasets improve fairness and predictive performance.

## Steps Applied
- Lowercasing
- Stopword Removal
- Tokenization
- Lemmatization

## Why?
Text preprocessing removes noise and converts raw text into structured format suitable for ML models.

```python
import pandas as pd

df_text = pd.DataFrame({
    "review": [
        "This movie was absolutely amazing and wonderful!",
        "The film was boring and too long.",
        "Excellent acting but the story was weak.",
        "I did not like this movie at all."
    ]
})

df_text
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))

df_text["tokens"] = df_text["tokens"].apply(
    lambda x: [word for word in x if word not in stop_words]
)

df_text
```

- Reviews were tokenized into individual words.
- Stopwords were removed to eliminate noise.

# Task 13: Time-Series Preprocessing – Air Quality Dataset

## Task Overview
The objective of this task is to preprocess time-series data by handling timestamps and smoothing.

## Steps Applied
- Datetime conversion
- Resampling
- Handling missing values
- Rolling mean smoothing

## Why?
Time-series preprocessing ensures consistent time intervals,
reduces noise, and improves modeling performance.

```python
import pandas as pd
import numpy as np

# Create hourly datetime range
date_range = pd.date_range(start="2023-01-01", periods=200, freq="H")

df_air = pd.DataFrame({
    "datetime": date_range,
    "PM2.5": np.random.normal(50, 10, 200)
})

# Introduce missing values
df_air.loc[df_air.sample(frac=0.1).index, "PM2.5"] = np.nan

df_air.head()
```

```python
df_air["datetime"] = pd.to_datetime(df_air["datetime"])
df_air.set_index("datetime", inplace=True)

df_air.head()
```

```python
df_air["PM2.5"] = df_air["PM2.5"].interpolate()

df_air.isnull().sum()
```

```python
df_daily = df_air.resample("D").mean()

df_daily.head()
```

```python
df_air["PM2.5_smooth"] = df_air["PM2.5"].rolling(window=5).mean()

df_air.head()
```

## Results & Observations

- Datetime column was converted and set as index.
- Missing values were handled using interpolation.
- Data was resampled to daily averages.
- Rolling mean smoothing reduced short-term fluctuations.
- Time-series data is now structured and ready for forecasting models.
